```python

# 📦 Setup: Cargar datos y preparar variables
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
import joblib

# Cargar CSV
df = pd.read_csv("AnexoET_RoundWinner_Limpio.csv")

# Variables seleccionadas
features = ['RoundHeadshots', 'TeamStartingEquipmentValue', 'PrimaryAssaultRifle', 'TimeAlive']
X = df[features]
y = df['RoundWinner']

# Separar y escalar
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)
```

```python

# 🔍 Logistic Regression
from sklearn.linear_model import LogisticRegression

model_logreg = LogisticRegression()
model_logreg.fit(X_train, y_train)

accuracy = accuracy_score(y_test, model_logreg.predict(X_test))
print("Precisión de Logistic Regression:", round(accuracy * 100, 2), "%")

joblib.dump(model_logreg, "Modelo_LogisticRegression.pkl")
```

```python

# 🔍 Decision Tree
from sklearn.tree import DecisionTreeClassifier

model_tree = DecisionTreeClassifier(max_depth=5, random_state=42)
model_tree.fit(X_train, y_train)

accuracy = accuracy_score(y_test, model_tree.predict(X_test))
print("Precisión de Decision Tree:", round(accuracy * 100, 2), "%")

joblib.dump(model_tree, "Modelo_DecisionTree.pkl")
```

```python

# 🔍 Random Forest
from sklearn.ensemble import RandomForestClassifier

model_rf = RandomForestClassifier(n_estimators=150, max_depth=5, random_state=42)
model_rf.fit(X_train, y_train)

accuracy = accuracy_score(y_test, model_rf.predict(X_test))
print("Precisión de Random Forest:", round(accuracy * 100, 2), "%")

joblib.dump(model_rf, "Modelo_RandomForest.pkl")
```

```python

# 🔍 KNN
from sklearn.neighbors import KNeighborsClassifier

model_knn = KNeighborsClassifier(n_neighbors=5)
model_knn.fit(X_train, y_train)

accuracy = accuracy_score(y_test, model_knn.predict(X_test))
print("Precisión de KNN:", round(accuracy * 100, 2), "%")

joblib.dump(model_knn, "Modelo_KNN.pkl")
```

```python

# 🔍 SVM
from sklearn.svm import SVC

model_svm = SVC(kernel='rbf', probability=True, random_state=42)
model_svm.fit(X_train, y_train)

accuracy = accuracy_score(y_test, model_svm.predict(X_test))
print("Precisión de SVM:", round(accuracy * 100, 2), "%")

joblib.dump(model_svm, "Modelo_SVM.pkl")
```
