```python
# 📦 Setup común: cargar datos, limpiar RoundWinner y preparar variables
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
import joblib

# Cargar CSV
df = pd.read_csv("Anexo ET_demo_round_traces_2022 (2).csv", sep=';', engine='python')
df = df[df['RoundWinner'].isin(['True', 'False'])].copy()
df['RoundWinner'] = df['RoundWinner'].map({'True': 1, 'False': 0})

features = [
    'RoundKills', 'RoundHeadshots', 'TeamStartingEquipmentValue',
    'RLethalGrenadesThrown', 'RoundStartingEquipmentValue',
    'PrimaryAssaultRifle', 'TimeAlive'
]

# Limpiar columnas numéricas
for col in features:
    df[col] = (
        df[col].astype(str)
        .str.replace('.', '', regex=False)
        .str.replace(',', '.', regex=False)
        .astype(float)
    )

df_model = df[features + ['RoundWinner']].dropna()
X = df_model[features]
y = df_model['RoundWinner']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42)
```

```python
# 🔍 Modelo: Logistic Regression
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)

accuracy = accuracy_score(y_test, model.predict(X_test))
print("Precisión de Logistic Regression: ", round(accuracy * 100, 2), "%")

joblib.dump(model, "Logistic_Regression_RoundWinner_Model.pkl")
```

```python
# 🔍 Modelo: Decision Tree
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(max_depth=5, random_state=42)
model.fit(X_train, y_train)

accuracy = accuracy_score(y_test, model.predict(X_test))
print("Precisión de Decision Tree: ", round(accuracy * 100, 2), "%")

joblib.dump(model, "Decision_Tree_RoundWinner_Model.pkl")
```

```python
# 🔍 Modelo: Random Forest
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=150, max_depth=5, random_state=42)
model.fit(X_train, y_train)

accuracy = accuracy_score(y_test, model.predict(X_test))
print("Precisión de Random Forest: ", round(accuracy * 100, 2), "%")

joblib.dump(model, "Random_Forest_RoundWinner_Model.pkl")
```

```python
# 🔍 Modelo: KNN
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train, y_train)

accuracy = accuracy_score(y_test, model.predict(X_test))
print("Precisión de KNN: ", round(accuracy * 100, 2), "%")

joblib.dump(model, "KNN_RoundWinner_Model.pkl")
```

```python
# 🔍 Modelo: SVM
from sklearn.svm import SVC

model = SVC(kernel='rbf', probability=True, random_state=42)
model.fit(X_train, y_train)

accuracy = accuracy_score(y_test, model.predict(X_test))
print("Precisión de SVM: ", round(accuracy * 100, 2), "%")

joblib.dump(model, "SVM_RoundWinner_Model.pkl")
```
